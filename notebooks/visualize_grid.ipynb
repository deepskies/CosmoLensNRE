# Visualizing the Parameter Space Grid

This notebook plots the ($\Omega_m$, $w_0$) parameter space showing the layout of the 9 test regions used for evaluating the NRE. 

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.ticker import AutoMinorLocator, FuncFormatter, MultipleLocator
```

```python
best_style = {
    "font.family": "sans-serif",
    "mathtext.fontset": "custom",
    "mathtext.rm": "TeX Gyre Heros",
    "mathtext.bf": "TeX Gyre Heros:bold",
    "mathtext.sf": "TeX Gyre Heros",
    "mathtext.it": "TeX Gyre Heros:italic",
    "mathtext.tt": "TeX Gyre Heros",
    "mathtext.cal": "TeX Gyre Heros",
    "mathtext.default": "regular",
    "figure.figsize": (10.0, 10.0),
    "font.size": 26,
    "axes.labelsize": "medium",
    "axes.unicode_minus": False,
    "xtick.labelsize": "small",
    "ytick.labelsize": "small",
    "legend.fontsize": "small",
    "legend.handlelength": 1.5,
    "legend.borderpad": 0.5,
    "xtick.direction": "in",
    "xtick.major.size": 12,
    "xtick.minor.size": 6,
    "xtick.major.pad": 6,
    "xtick.top": True,
    "xtick.major.top": True,
    "xtick.major.bottom": True,
    "xtick.minor.top": True,
    "xtick.minor.bottom": True,
    "xtick.minor.visible": True,
    "ytick.direction": "in",
    "ytick.major.size": 12,
    "ytick.minor.size": 6.0,
    "ytick.right": True,
    "ytick.major.left": True,
    "ytick.major.right": True,
    "ytick.minor.left": True,
    "ytick.minor.right": True,
    "ytick.minor.visible": True,
    "grid.alpha": 0.8,
    "grid.linestyle": ":",
    "axes.linewidth": 2,
    "savefig.transparent": False,
}
plt.style.use(best_style)
cols = ["#5790fc", "#f89c20", "#e42536", "#964a8b", "#9c9ca1", "#7a21dd"]
#set cols as the matplotlib default color cycle
plt.rcParams['axes.prop_cycle'] = plt.cycler(color=cols)
```

```python
# 9 test regions arranged in a 3x3 grid across (Om, w) space.
# Each colored box (0.15 x 0.2 in parameter space) defines one test region;
# the dashed rectangle encloses all 9 and marks the full test prior extent.
fig, ax = plt.subplots(1, 1, figsize=(8, 8), dpi=100)
ax.set_xlim(0.1, 0.9)
ax.set_ylim(-1.8, -0.35)
centers = [(0.3, -1.6), (0.3, -1.0), (0.3, -0.6),
           (0.5, -1.6), (0.5, -1.0), (0.5, -0.6),
           (0.7, -1.6), (0.7, -1.0), (0.7, -0.6)]
for i, center in enumerate(centers):
    rect = plt.Rectangle((center[0] - (0.15/2), center[1] - 0.1), 0.15, 0.2, linewidth=2, edgecolor=cols[i], facecolor='none')
    # rect2 = plt.Rectangle((center[0] - 0.1, center[1] - (0.25/2)), 0.2, 0.25, linewidth=2, edgecolor="grey", facecolor='none')
    ax.add_patch(rect)
    # ax.add_patch(rect2)
    # add text "Region i" at the center of each box
    ax.text(center[0], center[1], f"Region {i+1}", fontsize=12, ha='center', va='center')

# aligh the dashed region such that it tightly encloses all 9 boxes
rect_dashed = plt.Rectangle((0.225, -1.7), 0.55, 1.2, linewidth=2, edgecolor='black', facecolor='none', linestyle='--')
ax.add_patch(rect_dashed)
# add text "Entire Test data" inside the dahsed rectangle
ax.text(0.5, -1.3, "Entire Test prior", fontsize=18, ha='center', va='center')

# add text "Train prior" at the top centre of the plot
ax.text(0.5, -0.48, "Train prior", fontsize=18, ha='center', va='bottom')

ax.tick_params(axis='both', which='major', length=10, width=2)
ax.tick_params(axis='both', which='minor', length=6, width=2)
# rotate xtick labels
ax.tick_params(axis='x', rotation=45)
# show more minor ticks with labels
ax.xaxis.set_minor_locator(AutoMinorLocator())
ax.yaxis.set_minor_locator(AutoMinorLocator())
ax.xaxis.set_major_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(0.2))
ax.set_xlabel(r'$\Omega_m$')
ax.set_ylabel(r'$w$')
# plt.show()
plt.savefig('w_om_grid_regions.pdf', bbox_inches='tight', dpi=300)
```

```python
# print the boundaries of the boxes and the corrsponding grey boxes surrounding them as region 1 through region 9
for i, center in enumerate(centers):
    x_min = center[0] - (0.15 / 2)
    x_max = center[0] + (0.15 / 2)
    y_min = center[1] - 0.1
    y_max = center[1] + 0.1
    grey_x_min = center[0] - 0.1
    grey_x_max = center[0] + 0.1
    grey_y_min = center[1] - (0.25 / 2)
    grey_y_max = center[1] + (0.25 / 2)
    print(f"Region {i+1}: Train data label=1: ({x_min:.2f}, {y_min:.2f}) to ({x_max:.2f}, {y_max:.2f}), label-0: ({grey_x_min:.2f}, {grey_y_min:.2f}) to ({grey_x_max:.2f}, {grey_y_max:.2f})")
```
